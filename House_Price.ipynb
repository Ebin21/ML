```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=0)
print(X_train.shape, X_test.shape, Y_train.shape, Y_test.shape)
```

```python
Clf_LR = LogisticRegression()
Clf_LR.fit(X_train, Y_train)
```

```python
Y_test_pred = Clf_LR.predict(X_test)
Y_test_pred
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
confusion_matrix(Y_test, Y_test_pred)
```

```python
accuracy_score(Y_test, Y_test_pred) 
```

```python
from sklearn import preprocessing
scaler = preprocessing.StandardScaler().fit(X_train)
X_train_scaled = scaler.transform(X_train)
```

```python
scaler = preprocessing.StandardScaler().fit(X_test)
X_test_scaled = scaler.transform(X_test)
X_test_scaled
```

```python
from sklearn.neighbors import KNeighborsClassifier
CLR_KNN = KNeighborsClassifier(n_neighbors=1)   
CLR_KNN.fit(X_train_scaled, Y_train)
```

```python
confusion_matrix(Y_test, CLR_KNN.predict(X_test_scaled))
```

```python
accuracy_score(Y_test, CLR_KNN.predict(X_test_scaled))
```

```python
CLR_KNN_3 = KNeighborsClassifier(n_neighbors=3)   
CLR_KNN_3.fit(X_train_scaled, Y_train)
```

```python
accuracy_score(Y_test, CLR_KNN_3.predict(X_test_scaled))
```

```python
from sklearn.model_selection import GridSearchCV
param_grid = {'n_neighbors': np.arange(1, 10)}
grid_search_v= GridSearchCV(KNeighborsClassifier(), param_grid)
grid_search_v.fit(X_train_scaled, Y_train)
```

```python
grid_search_v.best_params_
```

```python
optimized_knn = grid_search_v.best_estimator_
confusion_matrix(Y_test, Y_test_pred)
```

```python
accuracy_score(Y_test, Y_test_pred)
```

```python
accuracy_score(Y_test, optimized_knn.predict(X_test_scaled))
```
